```python

import random


words = ["flood", "game" , "pro" , "Zombi" ,"number" , "home" , "music"]
secret = random.choice(words)
dlina = len(secret)
procherki = ["_"] * dlina
osibka = 6

print("Приветствую вас в игре 'Виселица'!")

while osibka > 0:
  print(f"Слово: {procherki}")
  print(f"Осталось жизней:{osibka}")

  vvedite = input("Введите ОДНУ букву: ")

  if vvedite in secret:
    print("Правильно! Такая буква есть.")
    for i in range(dlina):
      if secret[i] == vvedite:
          procherki[i] = vvedite
  else:
    print("Мимо! Такой буквы нет.")
    osibka -= 1
```
